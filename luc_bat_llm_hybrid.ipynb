```python
from hybrid_llm_generator import LLMDraftGenerator, RuleRepairEngine
from luc_bat_rules import check_luc_bat_poem_rules

draft_gen = LLMDraftGenerator()
repair_engine = RuleRepairEngine()

print("✓ Khởi tạo thành công Tầng 1 (LLM Draft) & Tầng 2 (Rule Repair Engine)!")
```

## 🧪 Demo Thực Nghiệm: Trước vs Sau Khi Sửa Lỗi Tự Động

```python
prompt = "nắng"
raw_draft = draft_gen.generate_draft(prompt)

print("=== TẦNG 1: BẢN THẢO RAW TỪ LLM (Có Lỗi Thừa/Thiếu Từ & Lệch Vần) ===")
for line in raw_draft:
    print(" ", " ".join(line).capitalize(), f"({len(line)} từ)")

raw_eval = check_luc_bat_poem_rules(raw_draft)
print(f"\n==> Đánh Giá RAW: {'✓ Đúng Luật' if raw_eval['valid'] else '✗ Lỗi Luật Thơ'}")
if not raw_eval['valid']:
    for err in raw_eval['errors']:
        print(f"   - {err}")
```

```python
repaired_poem = repair_engine.repair_poem(raw_draft)

print("=== TẦNG 2: BÀI THƠ SAU KHU ĐƯỢC RULE REPAIR ENGINE SỬA TỰ ĐỘNG ===")
for line in repaired_poem:
    print(" ", " ".join(line).capitalize(), f"({len(line)} từ)")

final_eval = check_luc_bat_poem_rules(repaired_poem)
print(f"\n==> Đánh Giá Sau Sửa: {'✓ THỎA MÃN 100% QUY TẮC LỤC BÁT' if final_eval['valid'] else '✗ Có Lỗi'}")
```
